```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
CUSTOM_COLORS =  ["green", "blue", "orange", "purple", "brown"]
```

```python
lab_data = [70, 150, 142]  
lab_labels = ["Laser Light", "Flurescent Light", "Day Light"]

field_data = [47, 22, 216]  
field_labels = ["Laser Light", "Street Light", "Day Light"]


# Increase the font size
plt.rcParams['font.size'] = 20


fig, ax = plt.subplots(1, 2, figsize=(10, 5))
# Create the pie chart for lab data without labels
ax[0].pie(lab_data, autopct=lambda p: '{:.0f}'.format(p * sum(lab_data) / 100), startangle=90)
ax[0].set_title('Lab Light Sources')

# Create the pie chart for field data without labels
ax[1].pie(field_data, autopct=lambda p: '{:.0f}'.format(p * sum(field_data) / 100), startangle=90)
ax[1].set_title('Field Light Sources')

# Equal aspect ratio ensures that pie is drawn as a circle.
ax[0].axis('equal')  
ax[1].axis('equal')

# Place the legend at the bottom with labels
plt.legend(lab_labels + field_labels, loc='lower center', bbox_to_anchor=(-0.1, -0.17), ncol=3)

# Adjust the layout
plt.subplots_adjust(bottom=0.2)
plt.savefig('light_sources.svg')
```

```python
# Data

import numpy as np
import matplotlib.pyplot as plt

CUSTOM_COLORS =  ["orange", "purple","blue","green"]

resnet_dict = {
    "Adequate light + fr": [77,85,87,85],
    "Adequate light + ev": [76,85,83,86],
    "Low light + fr": [75,75,80,72],
    "Low light + ev": [68,73,80,85],
}
vgg_dict = {
    "Adequate light + fr": [84,85,88,89],
    "Adequate light + ev": [78,89,89,90],
    "Low light + fr": [80,81,74,80],
    "Low light + ev": [72,81,85,85],
}
efficientnet_dict = {
    "Adequate light + fr": [81,89,90,92],
    "Adequate light + ev": [80,89,93,93],
    "Low light + fr": [78,83,83,85],
    "Low light + ev": [70,82,92,91],
}
mobilenetv2_dict = {
    "Adequate light + fr": [83,89,89,91],
    "Adequate light + ev": [78,87,91,92],
    "Low light + fr": [80,83,85,85],
    "Low light + ev": [68,85,90,91],
}

data = [resnet_dict,vgg_dict,efficientnet_dict,mobilenetv2_dict]

input_size = {
    "Adequate light + fr": [32, 64, 128,224],
    "Adequate light + ev": [32, 64, 128,224],
    "Low light + fr": [32, 64, 128,224],
    "Low light + ev": [32, 64, 128,224]
}

# Parameters
subplots = 4
subtitles = ["Resnet34", "VGG16", "Efficientnet-b0", "Mobilenetv2"]
#title = "Classification Performance"

fig, axs = plt.subplots(1, subplots, figsize=(20, 5), sharey=True)
axs_flatten = axs.flatten()

for i in range(subplots):
    for j, (k, v) in enumerate(data[i].items()):
        if j > 1:
            markerr = '*'
        else:
            markerr = 'o'
        axs_flatten[i].plot(input_size[k], v, label=k, c=CUSTOM_COLORS[j], marker=markerr, markersize=10, linestyle='-', linewidth=2.5)
        axs_flatten[i].set_xlabel('Input Size',fontsize=15)
        if i == 0:
            axs_flatten[i].set_ylabel('Accuracy (%)',fontsize=15)
        axs_flatten[i].set_xticks(input_size[k])
        axs_flatten[i].set_title(f"{subtitles[i]}",fontsize=18)

#plt.suptitle(title, y=1.1)
plt.legend(loc='lower center', bbox_to_anchor=(-1.3, -0.3), ncol=len(data),fontsize=15)
plt.savefig('input_size.svg', bbox_inches='tight')
# plt.legend()
```
